```python
#Mapa de Correlaciones
corr = df.corr(numeric_only=True)
plt.figure(figsize=(10,10))
sns.heatmap(
    corr,
    annot=True,      # Muestra los valores
    fmt=".1f",       # 1 decimal
    cmap="coolwarm",
    center=0
)
plt.title("Mapa de Correlaciones")
plt.show()
```

###Boxplot

```python
# Defino la variable 'numeric_columns' filtrando las columnas numéricas de df
numeric_columns = df.select_dtypes(include=np.number).columns

# Excluyo columnas con valores NaN
valid_columns = [col for col in numeric_columns if df[col].notna().any()]

# Ahora ejecutamos el bucle para crear boxplot de todas las variables numericas validas
for col in valid_columns:
    plt.figure(figsize=(2,2))
    sns.boxplot(x=df[col])
    plt.title(col)
    plt.show()
```
